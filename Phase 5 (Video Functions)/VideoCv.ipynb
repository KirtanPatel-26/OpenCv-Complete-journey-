```python
# This code tells us how to open my laptops camera and closing it

import cv2

cap = cv2.VideoCapture(0)

while True:
    ret, frame = cap.read()

    if not ret:
        print("Error..! Could not read image..")
        break

    cv2.imshow("Webcam Feed", frame)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        print("Image quitting..")
        break

cap.release()
cv2.destroyAllWindows()
```

```python
# cv2.VideoWriter(filename,codec,fps,frame_size)
# Opening and Saving video

frame_width = int(camera.get(cv2.CAP_PROP_FRAME_WIDTH))
frame_height = int(camera.get(cv2.CAP_PROP_FRAME_HEIGHT))

codec = cv2.VideoWriter_fourcc(*'XVID')
recorder = cv2.VideoWriter("my_video.avi",codec,20,(frame_width,frame_height))

while True:
    success, image = camera.read()

    if not success:
        break

    recorder.write(image)
    cv2.imshow("Recording live", image)

    if cv2.waitKey(1) & 0xFF == ord('q'):
        break

camera.release()
recorder.release()
cv2.destroyAllWindows()
```

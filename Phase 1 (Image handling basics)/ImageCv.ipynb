```python
import cv2
```

```python
print(cv2.__version__)
```

```python
image = cv2.imread("lenovo.jpg")
```

```python
# Image is getting fetched

if image is None:
    print("Error: Image Not Found !!")
else:
    print("Image found !!")
```

```python
# Image is getting printed and opening

if image is not None:
    cv2.imshow("Image Showing", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Error : Image not found !!")
```

```python
# Image is getting saved.

if image is not None:
    success = cv2.imwrite("output.png",image)
    if success :
        print("Image saved successfully to 'output.png' !!")
    else:
        print("Failed to save the image !!")
else :
    print("Error: Image not found !!")
```

```python
# Image dimensioning (height,width,channels) function used is shape

if image is not None:
    h,w,c = image.shape
    print(f"Image loaded:\nHeight {h}\nWidth {w}\nChannels {c}")
else: 
    print("Couldn't load image.......")
```

```python
# Greyscale Conversion (Converting image to black and white)
if image is not None :
    gray = cv2.cvtColor(image , cv2.COLOR_BGR2GRAY)
    cv2.imshow("Grayscale Image", gray)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
else:
    print("Could not load image !!")
```

```python
if image is not None:
    success2 = cv2.imwrite("GreyOutput.png",gray)
    if(success2) :
        print("Image saved successfully !!")
    else:
        print("Could not load image !!")
else :
    print("Error: Image not found !!")
```

```python
##### image2 = cv2.imread("oracle_badge.png")

if image2 is not None:
    print("Image is found..!!")
else: 
    print("Error: Image not found..!!")
    exit()
    
gray2 = None

while True:
    print("\n===== OpenCV Image Menu =====")
    print("1. Show Original Image")
    print("2. Convert to Grayscale")
    print("3. Show Grayscale Image")
    print("4. Save Image")
    print("5. Find height,Width,Channel")
    print("6. Exit")

    choice = input("Enter your Choice: ")

    if choice == "1":
        cv2.imshow("Original Image", image2)
        cv2.waitKey(0)
        cv2.destroyAllWindows()

    elif choice == "2":
        gray2 = cv2.cvtColor(image2 , cv2.COLOR_BGR2GRAY)
        print("Successfully converted to Grayscale !!")

    elif choice == "3":
        if gray2 is not None :
            cv2.imshow("Gray image",gray2)
            cv2.waitKey(0)
            cv2.destroyAllWindows()
        else:
            print("Please convert the image to grayscale first!")
            
    elif choice == "4":
        if gray2 is not None :
            filename = input("Enter the name of file you want to save: ")

            if not filename.lower().endswith((".jpg",".jpeg",".png")):
                filename += ".jpg"

            cv2.imwrite(filename,gray2)
            print(f"Image saved successfully as {filename}")

        else:
            print("Error: Please convert image to grayscale first..")

    elif choice == "5":
        h,w,c = image2.shape
        print(f"Image loaded:\nHeight {h}\nWidth {w}\nChannels {c}")

    elif choice == "6":
        print("Program Ended.......")
        break

    else:
        print("Invalid choice! Please enter 1 to 6")
            
```

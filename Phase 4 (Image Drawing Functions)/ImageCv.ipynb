```python
import cv2
```

```python
image = cv2.imread("oracle_badge.png")
```

```python
# Drawing Line

if image is None:
    print("Oops! Image not found..")
else:
    print("Image loaded successfully!")

    pt1 = (50,100)
    pt2 = (300,100)

    color = (255,0,0)
    thickness = 4

    cv2.line(image,pt1,pt2,color,thickness)

    cv2.imshow("Line image",image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Drawing Rectangle
# -1 thickness for colouring whole

if image is None:
    print("Error! Image not found.")
else:
    pt1 = (50,50)
    pt2 = (250,200)

    color = (0,0,255)
    thickness = 4

    cv2.rectangle(image,pt1,pt2,color,thickness)

    cv2.imshow("Rectange drawn image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Drawimg Circle
# -1 thickness for colouring whole

if image is None:
    print("Error! Image not found.")
else:
    pt = (150,150)

    color = (255,0,0)
    thickness = -1

    cv2.circle(image,pt,50,color,thickness)

    cv2.imshow("Circle drawn image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```

```python
# Adding text

if image is None:
    print("Image not found !!")
else :
    print("Image loaded successfully..")
    cv2.putText(image,"Hello Python..", (50,300),cv2.FONT_HERSHEY_COMPLEX,1.2,(255,0,0),2)

    cv2.imshow("New image", image)
    cv2.waitKey(0)
    cv2.destroyAllWindows()
```
